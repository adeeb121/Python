24. Count the frequency of each character in a string.

```python
s = "Adeeb"
frequency = {}
for char in s:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

for char,count in frequency.items():
    print(f"{char} : {count}")
```

25. Count the frequency of each word in a sentence. 

```python
sentence = "python is easy and python is useful"

words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

32. Find duplicate elements in a list.  

```python
numbers = [10, 20, 30, 20, 40, 10, 50, 20]

duplicates = []

for num in numbers:
    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print("Duplicate elements:", duplicates)
```

33. Find common elements between three lists.  

```python
list1 = [10, 20, 30, 40, 50]
list2 = [20, 30, 50, 60]
list3 = [10, 20, 30, 50]

common = []

for num in list1:
    if num in list2 and num in list3:
        if num not in common:
            common.append(num)

print("Common elements:", common)
```

34. Find the most frequent element in a list.  

```python
numbers = [10, 20, 10, 30, 20, 10, 40]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

most_frequent = max(frequency, key=frequency.get)

print("Most frequent element:", most_frequent)
print("Frequency:", frequency[most_frequent])
```

35. Remove duplicate words from a sentence.  

```python
sentence = "python is easy and python is useful"

words = sentence.split()
unique_words = []

for word in words:
    if word not in unique_words:
        unique_words.append(word)

result = " ".join(unique_words)

print(result)
```

37. Create an inventory tracker. 

```python
inventory = {
    "Pen": {"quantity": 50, "price": 10},
    "Notebook": {"quantity": 30, "price": 50},
    "Pencil": {"quantity": 40, "price": 5}
}

# Display inventory
for item, details in inventory.items():
    print(
        item,
        "Quantity:", details["quantity"],
        "Price:", details["price"]
    )

# Add stock to an existing product
inventory["Pen"]["quantity"] += 10

print("Updated Pen quantity:",
      inventory["Pen"]["quantity"])
```

38. Create a shopping cart using a list of dictionaries.  

```python
cart = [
    {"name": "Shirt", "price": 500, "quantity": 2},
    {"name": "Shoes", "price": 1200, "quantity": 1},
    {"name": "Cap", "price": 200, "quantity": 3}
]
total = 0
for item in cart:
    item_total = item["price"] * item["quantity"]

    print(item["name"], "Subtotal:", item_total)

    total += item_total

print("Total bill:", total)
```

40. Create a contact management system. 

```python
contacts = {}
while True:
    print("\n1. Add Contact")
    print("2. Search Contact")
    print("3. Display All Contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter name: ")
        phone = input("Enter phone number: ")

        contacts[name] = phone
        print("Contact saved successfully")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in contacts:
            print("Phone:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        if contacts:
            for name, phone in contacts.items():
                print(name, ":", phone)
        else:
            print("No contacts available")

    elif choice == "4":
        print("Exiting contact manager")
        break

    else:
        print("Invalid choice")
```
