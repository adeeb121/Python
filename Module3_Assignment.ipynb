```python
students = [
    {
        "roll": 101,
        "name": "Adeeb",
        "marks": 116
    },
    {
        "roll": 102,
        "name": "Ayush",
        "marks": 101
    },
    {
        "roll": 103,
        "name": "Harshit",
        "marks": 97
    },
    {
        "roll": 104,
        "name": "Gaurav",
        "marks": 105
    },
    {
        "roll": 105,
        "name": "Aman",
        "marks": 95
    },
    {
        "roll": 106,
        "name": "Dakah",
        "marks": 96
    },
    {
        "roll": 107,
        "name": "Chandni",
        "marks": 101
    },
    {
        "roll": 108,
        "name": "Himanshi",
        "marks": 108
    },
    {
        "roll": 109,
        "name": "Nancy",
        "marks": 92
    },
    {
        "roll": 110,
        "name": "Imran",
        "marks": 120
    },
]
```

1. Display all students 

```python
for student in students:
    print(student["roll"], student["name"], student["marks"])
```

2. Find the student with the highest marks 

```python
max = students[0]
for student in students:
    if student["marks"] > max["marks"]:
        max = student
print(max["roll"],max["name"],max["marks"])
```

3. Find the student with the lowest marks 

```python
min = students[0]

for student in students:
    if student["marks"] < min["marks"]:
        min = student

print(min["roll"], min["name"], min["marks"])
```

4. Calculate average marks 

```python
sum = 0
for i in students:
    sum = sum + i["marks"]
average = sum/10
print(average)
```

5. Display students scoring above 80 

```python
for i in students:
    if i["marks"]>96:
        print(i["roll"],i["name"],i["marks"])
```

6. Search student by roll number 

```python
roll = int(input("Enter roll no: "))
for i in students:
    if i["roll"]==roll:
       print(i["roll"],i["name"],i["marks"])
```

7. Count how many students passed

```python
passed = 0
for i in students:
    if i["marks"]>48:
        passed+=1
print(passed)
```

8. Sort students by marks 

```python
students.sort(key=lambda x: x["marks"], reverse=True)
for i in students:
    print(i["roll"], i["name"], i["marks"])
```

9. Display names in alphabetical order

```python
names = []
for student in students:
    names.append(student["name"])
names.sort()
for i in names:
    print(i)
```

10. Display a summary

```python
print("total students:", len(students))
print("average marks:", average)

print("highest:")
print(max["name"], "-", max["marks"])

print("lowest:")
print(min["name"], "-", min["marks"])

print("passed:", passed)
print("failed:", len(students) - passed)

print("students above 80:")

for student in students:
    if student["marks"] > 80:
        print(student["name"])
```
