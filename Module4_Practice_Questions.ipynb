```python
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print(largest(10, 25, 15))
```

Q16. Count vowels in a string

```python
def count_vowels(text):
    count = 0

    for ch in text.lower():
        if ch in "aeiou":
            count += 1

    return count

print(count_vowels("Hello Adeeb"))
```

Q17. Check palindrome

```python
def palindrome(text):
    if text == text[::-1]:
        return True
    else:
        return False

print(palindrome("Friends"))
```

Q32. Factorial using recursion

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

Q33. Sum of numbers from 1 to n

```python
def total(n):
    if n == 0:
        return 0

    return n + total(n - 1)

print(total(5))
```

Q34. Fibonacci numbers using recursion

```python
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(7):
    print(fibonacci(i), end=" ")
```

Q35. Student Result System

```python
def get_marks():
    marks = []
    for i in range(1, 6):
        m = float(input(f"Enter subject {i} marks: "))
        marks.append(m)
    return marks

def calculate_total(marks):
    return sum(marks)

def calculate_percentage(marks):
    return calculate_total(marks) / len(marks)

def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 40:
        return "D"
    else:
        return "Fail"

def display_result(marks):
    total = calculate_total(marks)
    percentage = calculate_percentage(marks)
    grade = calculate_grade(percentage)

    print("Total Marks:", total, "/ 500")
    print("Percentage:", percentage, "%")
    print("Grade:", grade)

marks = get_marks()
display_result(marks)
```

Q36. Number Analyzer

```python
def input_numbers():
    numbers = list(map(int, input("Enter numbers: ").split()))
    return numbers

def calculate_sum(numbers):
    return sum(numbers)

def calculate_average(numbers):
    return sum(numbers) / len(numbers)

def find_maximum(numbers):
    return max(numbers)

def find_minimum(numbers):
    return min(numbers)

def count_even(numbers):
    count = 0
    for n in numbers:
        if n % 2 == 0:
            count += 1
    return count

def count_odd(numbers):
    count = 0
    for n in numbers:
        if n % 2 != 0:
            count += 1
    return count

numbers = input_numbers()

if len(numbers) > 0:
    print("Sum:", calculate_sum(numbers))
    print("Average:", calculate_average(numbers))
    print("Maximum:", find_maximum(numbers))
    print("Minimum:", find_minimum(numbers))
    print("Even Count:", count_even(numbers))
    print("Odd Count:", count_odd(numbers))
else:
    print("Please enter at least one number.")
```

Q37. Menu-Driven Calculator

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

def power(a, b):
    return a ** b

while True:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "6":
        print("Calculator closed")
        break

    if choice not in ("1", "2", "3", "4", "5"):
        print("Invalid choice")
        continue

    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    if choice == "1":
        print("Result:", add(a, b))
    elif choice == "2":
        print("Result:", subtract(a, b))
    elif choice == "3":
        print("Result:", multiply(a, b))
    elif choice == "4":
        print("Result:", divide(a, b))
    elif choice == "5":
        print("Result:", power(a, b))
```
